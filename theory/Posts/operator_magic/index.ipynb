## Abstract
本文面向纯态定义了一种magicness的测度，

本文定义了操作$U$的算符Magic (OSE)，并指出其有如下性质：

1. 具有子系统可加性，忠实且Clifford下不变
2. 反映了线路$U$中的T gate至少有多少个，且也构成$U$的unitary stabilizer nullity的下界
3. 解析计算了在XXZ模型的Hamitonian下，一个局域算符的OSE是如何随时间变化的
4. 解析计算了Haar random unitary的OSE.


## SRE and OSE

由于$\sum_{P}\text{tr}^2(O P) = 2^n\text{tr}(O^2)$，则函数$p_{\psi}(P) = [2^n\text{tr}(O^2)]^{-1} \text{tr}^2(O P)$是一个概率分布。

特别地，

-  对于一个纯态$\psi$，函数 $p_{\psi}(x) = \frac{1}{2^n} \text{tr}^2(\psi P_x)$ 是一个概率分布，且稳定子态均匀支持在$|S| = 2^n$个稳定子 $S$ 上。
-  对于$\mathrm{tr}(OO^\dagger) = d$的算符$O$，函数 $p_{O}(x) = \frac{1}{4^n} \text{tr}^2(O P_x)$ 是一个概率分布，且Pauli算符只在一个元素上有分布。

因此，该分布的碰撞概率清晰地描述了纯态的稳定子性质
我们也可以计算更高阶的碰撞，从而得到 Rényi-$k$ 稳定子熵。

::: {.callout-note icon="false"}
## Definition: Stabilizer Renyi Entropy 
纯态的$k$ stabilizer-Rényi Entropy定义为：
\begin{align*}
\mathrm{SRE}_k(\psi) &=\frac{1}{1 -k} \left[\ln \left(\sum_x p_{\psi}^k(x)\right) - \ln \left(\frac{1}{2^{kn}} \cdot 2^n\right)\right]= \frac{1}{1 - k} \ln \left[\frac{1}{2^n} \sum_{P \in \mathcal{P}_n} \text{tr}^{2k}(\psi P)\right].
\end{align*}
F-norm归一的算符$O$的$k$-Operator stabilizer Entropy定义为：
\begin{align*}
\mathrm{OSE}_k(O) &=\frac{1}{1 -k} \left[\ln \left(\sum_x p_{O}^k(x)\right)\right]= \frac{1}{1 - k} \ln \left[\frac{1}{4^{kn}} \sum_{P \in \mathcal{P}_n} \text{tr}^{2k}(OP)\right].
\end{align*}
:::

例如$k = 2$时，
\begin{align}
\mathrm{SRE}_2(\psi) &= -\ln\left[2^n\sum_x p_{\psi}^2(x)\right] &=-\ln \left[\frac{1}{2^n} \sum_{P \in \mathcal{P}_n} \text{tr}^{4}(\psi P)\right].
\end{align}

有时候考虑到$\ln$函数的平均会给Haar average带来很大的阻碍，需要引入线性化SRE：

$$
\begin{aligned}
\mathrm{SRE}_{\mathrm{lin}}(\psi) = 1 - d^{-1} \sum_{P} |\mathrm{tr}(P \psi)|^4 = 1 - e^{-\mathrm{SRE}_2(\psi)}
\end{aligned}
$$

可以证明纯态中SRE最小的就是稳定子态。

::: {.callout-important icon="false"}
## Lemma: Properties of Stabilizer Renyi Entropy 
对于纯态，$\mathrm{SRE}_{k},k>1$ 度量最小也只能取到0，且仅当 $|\psi\rangle$ 为纯稳定子态时才能取到这个最小。
:::

::: {.callout-note collapse="true"}
## Proof (点击展开)
若 $M_k = 0$，则 $\sum_{P \in \mathcal{P}_n} \text{tr}^{2k}(\psi P) = {2^n}$。
对于纯态，我们已知 $\sum_{P \in \mathcal{P}_n} \text{tr}^{2}(\psi P) = 2^n$。
又因为 $0\leq \text{tr}^{2}(\psi P) \leq 1$ 且 $k >1$，则有 $\text{tr}^{2k}(\psi P) \leq \text{tr}^{2}(\psi P)$，且当且仅当 $\text{tr}^{2}(\psi P) = 1$ 时取等号。
因此：$$\sum_{P \in \mathcal{P}_n} \text{tr}^{2}(\psi P) \geq \sum_{P \in \mathcal{P}_n} \text{tr}^{2k}(\psi P)$$且当且仅当 $\text{tr}^{2}(\psi P) = 1$ 时取等号。
要使 $M_k=0$，必有 $\text{tr}^{2}(\psi P) = 1$ 成立，这意味着 $\psi$ 是一个稳定子态。
:::

对于混合态，该度量将不再忠实。为了解决这个问题，我们将稳定子混合态集放宽为形式如下的混合态：$\chi= {\mathbb{I}}/{2^n} + 2^{-n} \sum_{P \in G} \phi_P P$。

## Nonstabilizing power
#### Definitions
所谓Unitary的Nonstabilizeing power，指的是从一个没有资源的状态出发，经过U的作用，可以增加多少资源。

::: {.callout-note icon="false"}
## Definition: Power of SRE and  OSE
给定Unitary $U$，其OSE/SRE power定义为：
\begin{align}
\mathcal{P}_{\mathrm{S},k}(U) := \frac{1}{|\mathrm{STAB}|} \sum_{{\psi} \in \mathrm{STAB}} \mathrm{SRE}_k(U\psi U^\dagger) \\
\mathcal{P}_{\mathrm{O},k}(U) := \frac{1}{|\mathcal{P}_n|} \sum_{P \in \mathcal{P}_n} \mathrm{OSE}_k(U^\dagger P U) 
\end{align}
:::

同理也有：$\mathcal{P}_{\mathrm{S},\mathrm{lin}}(U) =  \frac{1}{|\mathrm{STAB}|} \sum_{{\psi} \in \mathrm{STAB}} \mathrm{SRE}_{\mathrm{lin}}(U\psi U^\dagger),$ 利用$-\ln x$是凹函数，可以证明如下不等式：
$$
\begin{aligned}
\mathcal{P}_{\mathrm{S},2}(U) &:=  -\frac{1}{|\mathrm{STAB}|} \sum_{{\psi} \in \mathrm{STAB}}\ln \left[\frac{1}{2^n} \sum_{P \in \mathcal{P}_n} \text{tr}^{4}(U\psi U^\dagger P)\right]\\
&\geq  -\ln \left[ \frac{1}{|\mathrm{STAB}|} \sum_{{\psi} \in \mathrm{STAB}}\frac{1}{2^n} \sum_{P \in \mathcal{P}_n} \text{tr}^{4}(U\psi U^\dagger P)\right] = -\ln \left[ 1 - \mathcal{P}_{\mathrm{S},\mathrm{lin}}(U) \right]
\end{aligned}
$$

#### T gates and  SRE/OSE
一个不惊讶的结论是，$\mathcal{P}_{\mathrm{S},k}(U)$和$\mathcal{P}_{\mathrm{O},k}(U)$某种程度上都反映了线路$U$中含有的$T$ gate的多少。


## Average over Haar random unitary

对于 Haar 随机态分布，$\mathbb{E}_{\psi \sim \nu}{\mathrm{SRE}}_k(\psi) = n \ln 2$.

我们关心的均值$D^{-\beta}\sum_{P}\overline{\langle P\rangle^{\beta}}$可以利用[Weingarten代数](/notes/quantum_learning_theory/Weingarten_calculus.qmd)计算如下：
\begin{align}
D^{-\beta}\sum_{P}\mathbb{E}_{U \sim \mathrm{Haar}} \left[ \mathrm{tr}(U^\dagger OU P)\right]^\beta =D^{-\beta} \sum_{\pi,\sigma \in S_{\beta}} \mathrm{Wg}_{\pi,\sigma} \mathrm{tr}(O^{\otimes \beta} \mathbb{P}^{\dagger}_{\pi})\sum_{P}\mathrm{tr}(P^{\otimes \beta} \mathbb{P}_{\sigma})
\end{align}
由于Pauli算符的乘积又很好的性质，上述求和有解析表达式。
我们设置$O$为非单位元的Pauli算符，考虑到只有在 $\pi$ 的cycle均为偶数时，$\mathrm{tr}(O^{\otimes \beta} \mathbb{P}_{\pi})$ 才非0，因此记 $\pi \in S^+_{\beta}$为仅包含偶数长度循环的置换，利用 $O^2 = \mathbb{I}$，则 $\mathrm{tr}(O^{\otimes \beta} \mathbb{P}_{\pi}) = D^{c(\pi)} \delta_{\pi, S^+_{\beta}}$。对于Pauli的求和，计算如下：
\begin{align}
\sum_{P}\mathrm{tr}(P^{\otimes \beta} \mathbb{P}_{\sigma}) = \left\{\begin{aligned} &1 \cdot D^{c(\sigma)} + (D^2 - 1)D^{c(\sigma)}, \text{~if~}\sigma \in S^+_{\beta}\\ &D^{c(\sigma)}, \text{~if~} \sigma \notin S^+_{\beta}\end{aligned} \right.
\end{align}
